```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
import pandas as pd
import numpy as np

# Load the Iris dataset
iris = load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_df['species'] = iris.target

# Map species numbers to actual species names
iris_df['species'] = iris_df['species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

# A. Scatter plot: Petal length vs Petal width
plt.figure(figsize=(8, 6))
sns.scatterplot(data=iris_df, x='petal length (cm)', y='petal width (cm)', hue='species', palette='viridis')
plt.title('Scatter Plot of Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.legend(title='Species')
plt.show()

# B. Histograms for each attribute
iris_df.drop(columns='species').hist(bins=15, figsize=(10, 8), color='skyblue', edgecolor='black')
plt.suptitle('Histogram of Iris Dataset Attributes')
plt.tight_layout()
plt.show()

# C. Pie chart for species frequency
species_counts = iris_df['species'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(species_counts, labels=species_counts.index, autopct='%1.1f%%', colors=sns.color_palette('Set3', n_colors=3), startangle=140)
plt.title('Frequency of Each Flower Species')
plt.axis('equal')
plt.show()

# D. Pair plot of all attributes
sns.pairplot(iris_df, hue='species', palette='Set2')
plt.suptitle('Pair Plot of Iris Dataset Attributes', y=1.02)
plt.show()
```
