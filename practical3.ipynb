```python
import pandas as pd
import numpy as np
import seaborn as sns
%matplotlib inline
df = pd.read_csv(r'C:\Users\Administrator\Downloads\titanic.csv')
df.head()
# B. Check for duplicate records and missing values
df = df.drop_duplicates()
df['Age'].fillna(df['Age'].median(), inplace=True)
df.dropna(inplace=True)
# C. Calculate and display the total number of passengers who survived and those who did not
survived = df['Survived'].value_counts()
print("Survived:\n{survived}")
# D. Filter the DataFrame to select records of passengers under the age of 18
children_df = df[df['Age'] < 18]
print(children_df)
# E. Calculate the average age for passengers in each passenger class
average_age_per_class = df.groupby('Pclass')['Age'].mean()
print(f"Average age per passenger class:\n{average_age_per_class}")
# F. Create a new column "Family Size" representing the total number of family members
df['Family Size'] = df['SibSp'] + df['Parch'] + 1
# G. Calculate the correlation between age and fare attributes
correlation = df[['Age', 'Fare']].corr()
print(f"Correlation between Age and Fare:\n{correlation}")
# H. Create a contingency table showing counts based on survival status and passenger class
contingency_table = pd.crosstab(df['Survived'], df['Pclass'], margins=True, margins_name='Total')
print("Contingency table (Survival vs Passenger Class):\n{contingency_table}")
```
