```python
import numpy as np
from scipy import stats

array=np.random.randint(1,101,size=100)
minimum=np.min(array)
maximum=np.max(array)
mean=np.mean(array)
median=np.median(array)
std_dev=np.std(array)
unique_values,counts=np.unique(array,return_counts=True)
num_unique=len(unique_values)
most_frequent_values=unique_values[np.argmax(counts)]
count_of_unique_values = dict(zip(unique_values,counts))

print("array=",array)
print("minimum=",minimum)
print("maximum=",maximum)
print("mean=",mean)
print("median=",median)
print("standerd derivation=",std_dev)
print("no. of unique values=",num_unique)
print("count of unique values=",count_of_unique_values)
print("most frequent value=",most_frequent_values)


```

```python
import numpy as np
heights=np.array([165, 170, 175, 168, 172, 180, 160, 169, 176, 171, 174, 182, 158, 167, 173, 179, 163, 166, 177, 181])
weights=np.array([60, 65, 70, 75, 80, 85, 55, 58, 63, 68, 72, 77, 50, 62, 67, 74, 52, 57, 69, 73])
print("heights (cm):",heights)
print("weights (cm):",weights)

combined=np.column_stack((heights,weights))
print("combination=",combined)

mean_height=np.mean(heights)
mean_weight=np.mean(weights)
print("mean height:",mean_height)
print("mean weight:",mean_weight)

index_shortest=np.argmin(heights)
index_tallest=np.argmax(heights)
print("index of shortest individual=",index_shortest)
print("index of tallest individual=",index_tallest)

sorted_combined=combined[combined[:,0].argsort()]
print("sorted array based on height:",sorted_combined)

swapped_combined=combined[:,[1,0]]
print("array with swapped columns:",swapped_combined)

under_70kg = combined[combined[:,1]<70]
print("individuals with weight below 70kg:",under_70kg)
```
