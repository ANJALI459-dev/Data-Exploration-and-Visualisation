```python
import plotly.express as px
from sklearn.datasets import load_iris
import pandas as pd

# Load the Iris dataset
iris = load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_df['species'] = iris.target

# Map species numbers to actual species names
iris_df['species'] = iris_df['species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

# A. Scatter plot: Petal length vs Petal width
fig = px.scatter(iris_df, x='petal length (cm)', y='petal width (cm)', color='species', title="Petal Length vs Petal Width",
                 labels={'petal length (cm)': 'Petal Length (cm)', 'petal width (cm)': 'Petal Width (cm)'},
                 color_discrete_sequence=px.colors.qualitative.Set2)
fig.show()

# B. Histograms for each attribute
fig = px.histogram(iris_df, x='sepal length (cm)', color='species', title='Histogram of Sepal Length (cm)', nbins=15)
fig.show()

fig = px.histogram(iris_df, x='sepal width (cm)', color='species', title='Histogram of Sepal Width (cm)', nbins=15)
fig.show()

fig = px.histogram(iris_df, x='petal length (cm)', color='species', title='Histogram of Petal Length (cm)', nbins=15)
fig.show()

fig = px.histogram(iris_df, x='petal width (cm)', color='species', title='Histogram of Petal Width (cm)', nbins=15)
fig.show()

# C. Pie chart for species frequency
species_counts = iris_df['species'].value_counts()
fig = px.pie(names=species_counts.index, values=species_counts.values, title="Frequency of Each Flower Species")
fig.show()

# D. Pair plot of all attributes (using scatter_matrix)
fig = px.scatter_matrix(iris_df, dimensions=iris.feature_names, color="species",
                        title="Pair Plot of Iris Dataset Attributes", 
                        color_discrete_sequence=px.colors.qualitative.Set2)
fig.show()
```
